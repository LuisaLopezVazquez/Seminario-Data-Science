## Exploracion de datos

```python
#Importar las librerias
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings 
warnings.filterwarnings("ignore")
```

```python
# Cargar archivo excel
file_path = "Accidentes.xlsx"
excel_data = pd.ExcelFile(file_path)
```

```python
#Ver el valor de las hojas
excel_data.sheet_names
```

```python
# Leer hoja valor
valor_df = pd.read_excel(file_path, sheet_name="valor")
```

```python
# Filtrar solo los datos realcionados con "Accidentes"
accidentes_filtered_df = valor_df[valor_df["unidad_medida"]=="Accidentes"]
```

```python
accidentes_filtered_df.head(5)
```

```python
valor_df.info()
```

```python
valor_df.describe()
```

```python
# Excluir la fila correspondiente a "Estados Unidos Mexicanos"
accidentes_filtered_df = accidentes_filtered_df[accidentes_filtered_df['desc_entidad']!= 'Estados Unidos Mexicanos']
```

```python
# Vamos agrupar por la columna  "desc_entidad", que contiene el nombre de las entidades
accidentes_por_estado=(
    accidentes_filtered_df.groupby('desc_entidad')
    .sum(numeric_only=True) # Suma unicamente las columnas numericas dentro de cada grupo (años)
    .iloc[:, 4:-1] # Seleccionamos las columnas que representan los datos de los años
)
```

```python
# Calcular el total de accidentes por estado
total_accidentes_por_estado = accidentes_por_estado.sum(axis=1).sort_values(ascending=False)
```

```python
# Crear un grafico de barras para mostar el total de accidentes por estado
plt.figure(figsize=(12,8))
total_accidentes_por_estado.plot(kind='bar', color='skyblue', edgecolor='black')
#total_accidentes_por_estado.plot(kind='area', alpha=0.5, color='lightblue')
plt.title("Total de Accidentes por Estado", fontsize=16)
plt.xlabel("Estado", fontsize=14)
plt.ylabel("Total de Accidentes", fontsize=14)
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.show()
```

## Mostrar el total de accidentes por año

```python
# Seleccionar solo las columnas numericas que representan los años
year_columns = [col for col in accidentes_filtered_df.columns if col.isdigit()]
year_columns
```

```python
# Crear un data frame que contenga unicamente las columnas de los años
accidentes_por_año =accidentes_filtered_df[year_columns]
accidentes_por_año
```

```python
# Sumar los valores de cada columna para obtener el total de accidentes por año
total_de_accidentes_por_año = accidentes_por_año.sum(axis=0)
total_de_accidentes_por_año
```

```python
# Crear una grafica de barras
import matplotlib.cm as cm
plt.figure(figsize=(12,8))
colormap = cm.get_cmap('tab20', len(total_de_accidentes_por_año))
colors = [colormap(i) for i in range(len(total_de_accidentes_por_año))]

total_de_accidentes_por_año.plot(kind="bar", color=colors, edgecolor="black")

plt.title("Total de accidentes por año", fontsize=16)
plt.xlabel("Año", fontsize=14)
plt.ylabel("Total de accidentes", fontsize=14)
plt.xticks(rotation = 45, ha="right", fontsize=12)
plt.show()
```

## Realizar el modelo de prediccion

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
import numpy as np
```

```python
# Filtrar los datos relevantes para realizar el grafico

accidentes_data = valor_df[valor_df['indicador']== "Accidentes de tránsito terrestre en zonas urbanas y suburbanas"]

# Seleccionar las de los años

accidentes_data = accidentes_data.loc[:,'1997':'2023'].sum() # Sumar todos los accidentes

# Convertir el indeci de accidentes por año en un tipo de dato fecha
accidentes_data.index = pd.to_datetime(accidentes_data.index, format="%Y")

#Vamos validar que los datos  sean de tipo float
accidentes_data_series = accidentes_data.astype(float)

# Crear las caracteristicas de año para entrenar el modelo
data = pd.DataFrame({'Año':accidentes_data_series.index.year,
                     'Accidentes':accidentes_data_series.values, #Pasa los valores de los accidentes
                    })
```

```python
x = data[['Año']] # Selecciono el año como la variable indenpediente
y = data['Accidentes'] # Selecciono y como la variable dependeintes
```

```python
# Dividir los datos  en un conjunto de entrenamiento (80%) u uno de prueba de (20%)
X_train, X_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42, shuffle=False)

# n_estimators es el numero de arboles de desicion 
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
```

```python
# Entrear el modelo usando los datos
rf_model.fit(X_train,y_train)
```

```python
# Predecir el numero de accidentes para el año 2024
future_year = pd.DataFrame({"Año": [2024]})
forecast_value = rf_model.predict(future_year)[0]
```

```python
forecast_value
```

```python
# Generar las predicciones para los datos historicos 
predictions = rf_model.predict(x)
```

```python
predictions
```

```python
# Graficar los resultados
plt.figure(figsize=(12,8))
plt.plot(data["Año"],data["Accidentes"], label="Reales", color="blue")

#Mostrar la prediccion para el año 2024
plt.scatter([2024], [forecast_value], label="Prediccion 2024", color='orange', s=100, edgecolor='black', zorder=5)

#Linea de prediccion
plt.plot(data["Año"],predictions, label="Prediccion del modelo", linestyle='--', color='green')

plt.title("Prediccion de Accidentes para el año 2024", fontsize=16)

plt.xlabel("Año", fontsize=14)

plt.ylabel("Total de Accidentes", fontsize=14)

plt.xticks(rotation = 45, fontsize=12)
plt.yticks(fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import r2_score, mean_absolute_error
#Calcular las metricas para Random Forest
rf_r2 = r2_score(y,predictions) # mide la variablidad de los datos obejtivo los accidentes
rf_mae = mean_absolute_error(y,predictions)  # Error absoluto
```

```python
print(f"R:{rf_r2:.4f}")  # Ramdom Forest explica el 93.45 % lo cual representa un ajuste muy bueno
print(f"MAE:{rf_mae:.2f}")
```
