## Manejar valores faltantes en un conjunto de datos utilizando la tecnica de Imputacion de K-NN

<h1> <font color="#004D7F" size=4> Modulo 3.- Machine Learning </h1>

```python
# Importar biblotecas necesarias
import numpy as np # manejar datos numericos
import pandas as pd # Para crear y administrar DataFrame
from  sklearn.impute import KNNImputer # Para imputacion con KNN
import seaborn as sns # Para visualizacion de datos
import matplotlib.pyplot as plt # Crear graficos
```

```python
#Generamos un dataset sintetico con algunas celdas que contienen valores faltantes(NaN)
np.random.seed(42) # Establecer la semilla para reproducibilidad
data ={
    'Edad': [25,30,np.nan,35,40,np.nan,50,55],
    'Ingresos': [30000,40000,50000,60000,np.nan,80000,np.nan,120000],
    'Puntaje': [700,750,800,np.nan,850,900,np.nan,950]
}

df = pd.DataFrame(data) # Creamos el data frame con los datos generados
```

```python
print("Datos originales con valores faltantes:")
print(df)
```

```python
# Utilizamos un mapa de calor para identificar las celdas con valores faltantes.
sns.heatmap(df.isnull(), cbar=False, cmap="viridis")
plt.title("Mapa de valores faltantes")
plt.show()
```

```python
# Aplicar imputacion KNN
imputer = KNNImputer(n_neighbors=3, weights='uniform') # configuracion basica
```

```python
# Imputar los valores faltantes y generar un nuevo DataFrame sin valores faltantes
df_imputed = pd.DataFrame(imputer.fit_transform(df), columns= df.columns)
```

```python
print("Datos después de la imputacion con KNN:")
print(df_imputed)
```
