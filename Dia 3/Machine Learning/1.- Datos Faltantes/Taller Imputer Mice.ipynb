## Tecnica de imputacion MICE

```python
import pandas as pd
import numpy as np
from sklearn.experimental import enable_iterative_imputer # Habilitar IterativeImputer
from sklearn.impute import IterativeImputer
```

```python
#Crear el dataset
data = pd.DataFrame({
    'ID':[1,2,3,4,5],
    'Programacion':[85,78,np.nan,92,75],
    'Calculo': [90,np.nan,70,80,np.nan],
    'BaseDeDatos':[88,80,72,np.nan,78]
})
```

```python
print("Mostrar el dataset origianl")
print(data)
```

```python
#Configurar el imputador MICE
imputer = IterativeImputer(max_iter=10, random_state=42) 
```

```python
# Aplicar la tecnica de MICE
data_imputed = imputer.fit_transform(data.drop(columns=['ID']))
```

```python
# Convertir los datos imputados a un DataFrame
data_imputed = pd.DataFrame(data_imputed, columns=["Programacion","Calculo","BaseDeDatos"])
```

```python
#Agregergar la columna ID
data_imputed.insert(0,'ID', data['ID'])
```

```python
print("Datos imputados")
print(data_imputed)
```
