## Implementacion de One Hot Enconding en un data set relacionado a Recursos Humanos 

```python
#Importar las librerias
import pandas as pd
import numpy as np
```

```python
df_rh = pd.read_csv("HRDataset_v14.csv")
```

```python
df_rh.head()
```

```python
df_rh1 = df_rh[["Employee_Name","Position","State","Sex","DOB","MaritalDesc",
               "CitizenDesc","Department","PerformanceScore","EngagementSurvey","EmpSatisfaction"]]
```

```python
df_rh1.shape
```

```python
df_rh1.head()
```

```python
# Crear variables dummy para la columna categorica de Sex
dummy_sex= pd.get_dummies(df_rh1["Sex"], prefix="sex")
```

```python
dummy_sex.head()
```

```python
# Eliminar la columna de "Sex" del DataFrame original para evitar redundancia
df_rh1 = df_rh1.drop(["Sex"], axis=1)
```

```python
# Agrear columnas dummy al DataFrame original
df_rh2 = pd.concat([df_rh1,dummy_sex], axis=1)
df_rh2
```

```python
# Crear variables dummy para una columna en especifico
def createdummies(df,varname):
    #pd.get_dummies() de Pandas se utiliza para convertir variables categóricas 
    #en variables dummy o indicadores
    dummy = pd.get_dummies(df[varname], prefix= varname)
    # Eliminar la columna original para evitar redundancia
    df = df.drop(varname, axis=1)
    #Concatenar las columnas dummy con el dataframe original 
    df=pd.concat([df,dummy], axis=1)
    return df
createdummies(df_rh2, "MaritalDesc")
    
```
