```python
import numpy as np 
```

```python
array_uni = np.array([1,3,5,7,9,11,50])
print("Shape:", array_uni.shape)
print("Array_unidades:",array_uni )
```

```python
array_uni[4]
```

```python
array_uni[2:4]
```

```python
array_uni[0:3]
```

```python
# Crear una array de dos dimensiones 
array_multidimencional= np.array([[1,5,6,7], [9,10,15,16]])
print("Shape:", array_multidimencional.shape)
print("Array_unidades:",array_multidimencional )
```

```python
array_multidimencional[1,3]
```

```python
array_multidimencional[0,0:2]
```

```python
array_multidimencional_random_int = np.random.randint(0,1000, size=(3,4))
print("Shape:", array_multidimencional_random_int.shape)
print("Array_unidades:",array_multidimencional_random_int )
```

```python
array_3d_random_int = np.random.randint(0,1000, size=(3,4,5))
print("Shape:", array_3d_random_int.shape)
print("Array_unidades:",array_3d_random_int)
```
