Ejemplo sencillo:
Imagina que tienes una bolsa llena de naranjas. Cada naranja tiene un peso ligeramente diferente, pero la mayoría pesa alrededor de 150 gramos. Sin embargo, algunas son más pequeñas (130 g) y otras más grandes (170 g).

Si pesas todas las naranjas y haces un gráfico con los resultados, el gráfico tendrá forma de campana:

La mayoría de las naranjas estará cerca del peso promedio (150 g). Habrá pocas naranjas muy ligeras (menos de 130 g) o muy pesadas (más de 170 g). ¡Esa es la distribución normal! Es como una campana que muestra cómo se distribuyen los datos: muchos en el centro (lo más común) y menos en los extremos (lo menos común).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

```

```python
pesos = np.random.normal(150, 10, 1000)
```

```python
#Creamos un histograma para visualizar los datos
count, bins,_ = plt.hist(pesos, bins=30, density=True, color='orange', edgecolor="black", alpha=0.7)

#Generar la linia en forma de campana
media = 150 #Promedio
desviacion = 10
x = np.linspace(bins[0], bins[-1], 1000) # Valores para el eje x
y = norm.pdf(x,media,desviacion) # Calculamos la densidad de probabilidad
plt.plot(x,y, 'r-', linewidth=2, label="Campana de Gauss")

# Etiquetas de titulo
plt.title("Distribuccion de pesos de las Naranjas")
plt.xlabel("Peso (gramos)")
plt.ylabel("Densidad")

#Mostrar el grafico
plt.show()
```

```python
pesos_sesgo_izq = np.random.normal(150,10,1000) - np.random.exponential(10, 1000) # genera 1000 valores aleatorios siguiendo una distribución normal (gaussiana) con los siguientes parámetros Media (150): 

pesos_sesgo_der = np.random.normal(150,10,1000) + np.random.exponential(10, 1000) # genera 1000 valores aleatorios siguiendo una distribución normal (gaussiana) con los siguientes parámetros Media (150): 

plt.hist(pesos_sesgo_izq, bins=30, color="blue", edgecolor="black", density=True)
plt.title("Sesgo hacia la izquierda")
plt.xlabel("Peso en gramos")
plt.ylabel("Densidad")
plt.show()


plt.hist(pesos_sesgo_der, bins=30, color="green", edgecolor="black", density=True)
plt.title("Sesgo hacia la Derecha")
plt.xlabel("Peso en gramos")
plt.ylabel("Densidad")
plt.show()
```
