En este taller exploraremos las medidas de tendencia central y las medidas de dispersión utilizando los datos de una base de punto de venta. Usaremos Python para realizar los cálculos directamente desde la tabla "ventas" de la base de datos.

Parte 1: Configuración Inicial

Primero, importemos las bibliotecas necesarias y carguemos los datos:


```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Abrir Conexion a la base datos
conexion = sqlite3.connect('punto_venta.db')
```

```python
#Crear un cursor
cursor = conexion.cursor()
```

```python
query ="SELECT name FROM sqlite_master WHERE type='table';"

tables=pd.read_sql_query(query,conexion)
```

```python
print("Tablas disponibles:")
print(tables)
```

```python
# Cargar los datos en un DataFrame usando una consulta SQL
query = "SELECT monto,total_productos FROM  ventas" 
df = pd.read_sql_query(query,conexion)
df
```

```python

# Cálculo de medidas de tendencia central y dispersión usando funciones de pandas
# Medidas para "monto"
media_monto = df["monto"].mean()
mediana_monto = df["monto"].median()
moda_monto =df["monto"].mode()[0]
rango_monto = df["monto"].max() - df["monto"].min()
varianza_monto= df["monto"].var()
desviancion_estandar= df["monto"].std()
```

```python
# Mostrar resultados
print("Medidas estadisticas")
print(f"Media: {media_monto}")
print(f"Mediana: {mediana_monto}")
print(f"Moda: {moda_monto}")
print(f"Rango: {rango_monto}")
print(f"Varianza: {varianza_monto}")
print(f"Desvación Estándar: {desviancion_estandar}")
```

```python
# Medidas para "total_productos"
media_productos = df['total_productos'].mean()
mediana_productos = df['total_productos'].median()
moda_productos = df['total_productos'].mode()[0]
rango_productos = df['total_productos'].max() - df['total_productos'].min()
varianza_productos = df['total_productos'].var() #Qué tanto varian los totales de productos.
desviacion_productos = df['total_productos'].std() 
```

```python
print("\nMedidas para 'total_productos':")
print(f"Media: {media_productos}")
print(f"Mediana: {mediana_productos}")
print(f"Moda: {moda_productos}")
print(f"Rango: {rango_productos}")
print(f"Varianza: {varianza_productos}") #Qué tanto varía el número de productos vendidos por transacción.
print(f"Desviación Estándar: {desviacion_productos}") #En promedio, cuánto varía el número de productos respecto al promedio de productos por venta.
```

```python
# Cerrar la conexión
conexion.close()
```
